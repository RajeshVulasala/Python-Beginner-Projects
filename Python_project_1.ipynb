```python
# creating a student list

def add_student():

    s_id = int(input("Enter student ID:"))
    name = input("Enter Student name")

    py_marks = float(input("Enter the Pyhton marks"))
    while py_marks < 0 or py_marks > 100:
        print("Invalid Pyhton Markes!")
        py_marks = float(input("Enter the Pyhton marks Again"))

    St_marks = float(input("Enter the Statistics marks"))
    while St_marks < 0 or St_marks > 100:
        print("Invalid Statistics Marks, Check Once")
        St_marks = float(input("Enter the Statistics marks Again"))
    
    Sql_marks = float(input("Enter the SQL marks"))
    while Sql_marks < 0 or Sql_marks > 100:
        print("Invalid Sql Marks Check Once Again")
        Sql_marks = float(input("Enter the SQL marks"))

    # Calculating the marks
    total = py_marks + St_marks + Sql_marks
    avg = total / 3
    percentage = (total / 300) * 100
    
    # Grade Calculator
    if percentage >= 91 and total <= 100:
        grade = "A"
    elif percentage >= 81:
        grade = "B"
    elif percentage >= 71:
        grade = "C"
    elif percentage >= 61:
        grade = "D"
    else:
        grade = "Better Hard Work Next Time"


    # printing the results
    print()
    print("==========Student Report Card =============")
    print("Student ID   :", s_id)
    print("Student Name :", name)
    print("Pyhton       :", py_marks)
    print("Statistics   :", St_marks)
    print("SQL          :", Sql_marks, "\n")
    print("Total Marks  :", total) 
    print("Average Marks:", round(avg, 2))
    print("Percentage   :", round(percentage, 1),"%")
    print("Grade        :", grade) 
    
add_student()
```
